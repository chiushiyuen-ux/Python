```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
def NDD(x,y):
    coef=y.astype(float).copy()
    n=len(x)
    for i in range(1,n):
        coef[i:n]=(coef[i:n]-coef[i-1:n-1])/(x[i:n]-x[0:n-i])
    return coef

def NDDplot(x,y,h=0.01):
    '''x:1*N array
    y:1*N array'''
    a=x[0]
    b=x[-1]
    N=int((b-a)/h)
    ndd=NDD(x,y)
    
    xpts=np.linspace(a,b,N)
    ypts=np.zeros_like(xpts)
    for i,coef in enumerate(ndd):
        ypts+=coef*np.prod(xpts-x[:i,None],axis=0)
    
    
    fig,ax=plt.subplots()
    plt.plot(x,y,'o',label='data points')
    plt.plot(xpts,ypts,label='NDD interpolation')
    plt.legend()
    plt.show()
    return None
```

```python
x=np.array([1,2,3,4,5],dtype=float)
y=np.array([1,4,1,16,25],dtype=float)

ndd=NDD(x,y)
print(ndd)
```

```python
NDDplot(x,y)
```
