```python
import numpy as np
```

```python
def f(x):
    return np.sin(x)-0.5*x
```

```python
def bisect(xu,xd,maxit,tol):
    err,it=1,0
    while err>tol and it<maxit:
        it+=1
        xm=(xu+xd)/2
        #print(xm,f(xm))
        if f(xm)*f(xu)<0:
            xd=xm
            err=abs(xd-xu)
        else:
            xu=xm
            err=abs(xd-xu)
    return xm,f(xm),it,err
```

```python
xu,xd=1.5,2.0
maxit,tol=1000,1e-6
xm,ym,it,err=bisect(xu,xd,maxit,tol)
print('xm:',xm,'ym:',ym,'it:',it,'err:',err)
```
