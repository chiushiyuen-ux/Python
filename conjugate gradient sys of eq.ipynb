```python
import numpy as np
```

```python
def congrad(A,x0,f,tol=1e-6,maxit=1000):
    A=A.copy()
    dim=A.shape[0]
    x=x0.copy()
    f=f.copy()
    
    r=f-A@x
    p=r.copy()
    it=0
    
    while np.linalg.norm(r,np.inf)>tol and it<maxit:
        it+=1
        inter=A@p
        a=np.inner(r,r)/np.inner(p,inter)
        x+=a*p
        r_new=r-a*inter
        b=np.inner(r_new,r_new)/np.inner(r,r)
        p=r_new+b*p
        r=r_new.copy()
    
    return x
```

```python
#since A may not be pos def, I use A.T@A to be mat and A.T@f to be rhs, then congrad works

A=np.zeros((10,10))
for i in range(10):
    A[i,i]=2
    if i>0:
        A[i,i-1]=-1
    if i<9:
        A[i,i+1]=-1
A=A.T@A

x0=np.ones(10)

f=2*np.ones(10)
f=A.T@f

print(A)
print(f)
```

```python
x=congrad(A,x0,f)
print(x)
print(A@x)
```
