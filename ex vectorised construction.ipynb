```python
# some vectorisable objects
# most tasks requires at least one loop, especially for itertive methods in maths. 
# However, there are still some structured objects that does not require loops and can be vectorised.
# In python, the trick is Mat[None:]. This makes a new dimension for the matrix, and then we can use broadcasting to do the vectorised operation.
```

```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
# Vandermonde matrix (not talking about space complexity when compared with interpolation like NDD) can be constructed in O(1)
x=np.array([2,5,11,13],dtype=float)
N=len(x)
V=x[:,None]**np.arange(N)
print(V)
```

```python
# vector distance matrix 
x=np.arange(0,6,1,dtype=float)
D=abs(x[:,None]+x[None,:])
print(D)
```
