example [9,5,7,2,3,6]    Tree with maximum 2^n elements. Here 9>5,7>2,3,6

9 is called the root, and others are called heap.
swap root with last heap value=> [6,5,7,2,3,9] => root smaller than one of the heap, so exchange =>[7,5,6,2,3,9]
store the remaining: =>[7,5,6,2,3],[9]

swap root with last heap value=> [3,5,6,2,7],[9]
root smaller than one of the heap, so exchange =>[6,5,3,2,7],[9]
store the remaining: =>[6,5,3,2],[7,9]

swap root with last heap => [2,5,3,6],[7,9]
root smaller than heap, swap =>[5,2,3,6],[7,9]
store the remaining value: =>[5,2,3],[6,7,9]

swap the root with last heap => [3,2,5],[6,7,9]
store the remaining value: =>[3,2],[5,6,7,9]

swap the root with last heap => [2,3],[5,6,7,9]
store the remaining value: => [2],[3,5,6,7,9]

store the remaining value: => [2,3,5,6,7,9]

Done


example [2,3,5,6,7,9]

first heap smaller than second heap, swap: [2,7,9,6,3,5]
root heap smaller than heap, swap: [9,7,2,6,3,5]
first heap smaller than second heap, swap: [9,7,5,6,3,2]
