```python
import numpy as np
```

```python
def merge(lst1,lst2):
    res=np.zeros(len(lst1)+len(lst2))
    i=0
    j=0
    while i<len(lst1) and j<len(lst2):
        if lst1[i]<lst2[j]:
            res[i+j]=lst1[i]
            i+=1
        else:
            res[i+j]=lst2[j]
            j+=1
    res[i+j:]=np.concatenate((lst1[i:],lst2[j:]))
    return res
```

```python
def mergesort(lst):
    if len(lst)<=1:
        return lst
    mid=len(lst)//2
    left=mergesort(lst[:mid])
    right=mergesort(lst[mid:])
    return merge(left,right)
```

```python
lst=np.random.randint(0,10,20)
print(lst)
fin=mergesort(lst)
print(fin)
```
