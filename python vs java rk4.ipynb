```python
import numpy as np
import matplotlib.pyplot as plt
import pandas as pd
```

```python
def f(x,t):
    return 2*t+x**2

a,b,N= 0,3,100
h=(b-a)/N
tpts=np.linspace(a,b,N)
xpts=[0.5]
x0=0.5

for t in tpts[:-1]:
    k1=h*f(x0,t)
    k2=h*f(x0+0.5*k1,t+0.5*h)
    k3=h*f(x0+0.5*k2,t+0.5*h)
    k4=h*f(x0+k3,t+h)
    x0+=(k1+2*k2+2*k3+k4)/6
    xpts.append(x0)
```
