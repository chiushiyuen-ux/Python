```python
import numpy as np
```

```python
def f(x):
    return x**2 + 2*x + 1
```

```python
def Newton(f,x0,maxit=1000,tol=1e-6):
    h = 1e-5
    for i in range(maxit):
        f_prime = (f(x0 + h) - f(x0 - h)) / (2 * h)
        x1 = x0 - f(x0) / f_prime
        if abs(x1 - x0) < tol:
            return x1
        x0 = x1
    return x0

Newton(f, 200)
```

```python
def Secant(f,x0,x1,maxit=1000,tol=1e-6):
    for i in range(maxit):
        f_x0 = f(x0)
        f_x1 = f(x1)
        x2 = x1 - f_x1 * (x1 - x0) / (f_x1 - f_x0)
        if abs(x2 - x1) < tol:
            return x2
        x0, x1 = x1, x2
    return x1

Secant(f, 200, 201)
```
